# Test of CaSR Forcing Fields against HRDPS

After correcting the units, the scripts reexamines the CaSR units.



```python
# visualize all kinds of variables to see if there are
# any accumulation/flux mistakes

import xarray as xr
import numpy as np

# Define file paths
path_CaSR = '/ocean/jqiu/forcings/CaSR/CaSR_y2023m03d02.nc'
path_HRDPS = '/results/forcing/atmospheric/continental2.5/nemo_forcing/hrdps_y2023m03d02.nc'

# load data

ds_CaSR=xr.load_dataset(path_CaSR)
ds_HRDPS=xr.load_dataset(path_HRDPS)


# grid number
# CaSR: y = 79, x = 69
# HRDPS: y = 229, x = 189

# list of all variables

list_var=['atmpres', 'precip', 'qair', 'tair', 'therm_rad']

# visualize

import matplotlib.pyplot as plt

row_number=len(list_var)

fig, axes = plt.subplots(row_number,1,
    figsize=(12, 18),
    constrained_layout=True,
)

# use loop to read the variable

for i in range(row_number):
    var=list_var[i]
    series_CaSR=ds_CaSR[var].isel(y=79,x=69).values
    series_HRDPS=ds_HRDPS[var].isel(y=229,x=189).values

    axes[i].plot(series_CaSR, label='CaSR')
    axes[i].plot(series_HRDPS, label='HRDPS')
    axes[i].set_title(var)
    axes[i].legend()
```
