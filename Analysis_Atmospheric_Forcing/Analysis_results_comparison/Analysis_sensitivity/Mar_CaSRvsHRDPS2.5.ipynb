### CaSR wind + HRDPS else

```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.colors import TwoSlopeNorm
from pathlib import Path

# ============================================================
# 1. USER SETTINGS
# ============================================================

MONTH = "Mar"
EXPERIMENT = "HRDPS2.5+CaSR_wind"       # HRDPS1 / CaSR / HRDPS10 / "HRDPS2.5+CaSR_wind" / "CaSR+HRDPS2.5_wind"
REFERENCE = "HRDPS2.5"

VARIABLE = "diatoms"       # votemper / vosaline / nitrate 
DATA_TYPE = "Biological"      # Physical / Biological

# Start with one station; add more stations later.
SELECTED_POINTS = [
    # "Central SoG",
    # "Juan de Fuca",
    # "Haro Strait",
    # "Northern SoG",
    "Puget Sound",
    "Pam Rocks",
    "SandHeads",
]

# Color settings for the two experiments
CMAP = "viridis"
VMIN = None                 # e.g. 6
VMAX = None                 # e.g. 10

# Difference = experiment - reference
DIFF_CMAP = "RdBu_r"
DIFF_MAX = None             # e.g. 1.0, gives [-1, 1]

# ------------------------------------------------------------
# Contour levels: manually fill these in.
# None = no contour lines
# ------------------------------------------------------------

# CONTOUR_LEVELS = None
# Example:
CONTOUR_LEVELS = np.arange(0, 8, 2)
# CONTOUR_LEVELS = [20, 24, 28, 29, 30, 30.2, 30.4]

# DIFF_CONTOUR_LEVELS = None
# Example:
# DIFF_CONTOUR_LEVELS = np.arange(-1.0, 1.01, 0.25)
DIFF_CONTOUR_LEVELS = np.arange(-4, 4, 1)

SHOW_CONTOUR_LABELS = True
CONTOUR_COLOR = "white"
DIFF_CONTOUR_COLOR = "black"
CONTOUR_LINEWIDTH = 0.6

# Units (change as needed)
# UNITS = {
#     "votemper": "°C",
#     "vosaline": "g kg$^{-1}$",
#     "nitrate": "mmol m$^{-3}$",
#     "diatoms": "mmol N m$^{-3}$",
#     "flagellates": "mmol N m$^{-3}$",
# }


# ============================================================
# 2. LOAD DATA
# ============================================================

def get_file_path(experiment, month, data_type):
    prefix = EXPERIMENTS_head[experiment][month]
    suffix = EXPERIMENTS_tail[data_type]
    path = Path(prefix + suffix)

    if not path.is_file():
        raise FileNotFoundError(f"File not found: {path}")

    return path


def load_dataset(experiment, month, data_type):
    path = get_file_path(experiment, month, data_type)
    print(f"Opening: {path}")
    return xr.open_dataset(path)


# ============================================================
# 3. EXTRACT TIME-DEPTH DATA AT ONE GRID POINT
# ============================================================

def extract_time_depth(ds, variable, station):
    """
    Return a DataArray with dimensions:
        (time_counter, deptht)

    The original variable is expected to have dimensions:
        (time_counter, deptht, y, x)
    """

    y, x = POINTS[station]
    #n_layers = Layers[station]
    n_layers = min(20, Layers[station])

    da = ds[variable].isel(y=y, x=x)

    # Limit to valid depth levels.
    da = da.isel(deptht=slice(0, n_layers))

    # Apply ocean mask.
    point_mask = tmask.isel(y=y, x=x)

    # NEMO mesh masks often have a singleton time dimension.
    if "t" in point_mask.dims:
        point_mask = point_mask.isel(t=0)
    if "time_counter" in point_mask.dims:
        point_mask = point_mask.isel(time_counter=0)

    # Rename mask depth dimension if necessary.
    if "z" in point_mask.dims:
        point_mask = point_mask.rename({"z": "deptht"})

    point_mask = point_mask.isel(
        deptht=slice(0, n_layers)
    )

    # Use matching depth coordinates to avoid xarray misalignment.
    point_mask = point_mask.assign_coords(
        deptht=da.deptht
    )

    da = da.where(point_mask == 1)

    # Remove missing and invalid values.
    da = da.where(np.isfinite(da))

    # Ensure time-depth dimension order.
    da = da.transpose("time_counter", "deptht")

    return da.load()


# ============================================================
# 4. ALIGN TWO EXPERIMENTS
# ============================================================

def align_experiments(ref_da, exp_da):
    """
    Match common dates and depth levels.

    Useful because HRDPS1 March ends on March 28,
    while HRDPS2.5 March ends on March 31.
    """

    ref_da, exp_da = xr.align(
        ref_da,
        exp_da,
        join="inner"
    )

    if ref_da.sizes["time_counter"] == 0:
        raise ValueError("No overlapping time coordinates.")

    if ref_da.sizes["deptht"] == 0:
        raise ValueError("No overlapping depth coordinates.")

    diff_da = exp_da - ref_da

    return ref_da, exp_da, diff_da


# ============================================================
# 5. PLOT SINGLE TIME-DEPTH PANEL
# ============================================================

def plot_time_depth(
    ax,
    da,
    cmap,
    vmin,
    vmax,
    contour_levels=None,
    norm=None,
    contour_color="white"
):
    time = pd.to_datetime(da.time_counter.values)
    depth = da.deptht.values
    values = da.transpose("deptht", "time_counter").values

    # Pseudocolor (time-depth field)
    mesh = ax.pcolormesh(
        time,
        depth,
        values,
        shading="auto",
        cmap=cmap,
        vmin=vmin if norm is None else None,
        vmax=vmax if norm is None else None,
        norm=norm,
        rasterized=True
    )

    # Contour lines
    if contour_levels is not None and len(time) >= 2:
        cs = ax.contour(
            time,
            depth,
            values,
            levels=contour_levels,
            colors=contour_color,
            linewidths=CONTOUR_LINEWIDTH
        )

    if SHOW_CONTOUR_LABELS:
        ax.clabel(
            cs,
            inline=True,
            fontsize=7,
            fmt="%g",
            colors=contour_color
        )
    # Depth increases downward.
    ax.invert_yaxis()

    ax.xaxis.set_major_locator(
        mdates.AutoDateLocator(minticks=3, maxticks=7)
    )
    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%b %d")
    )

    ax.tick_params(axis="both", labelsize=9)
    ax.grid(False)

    return mesh


# ============================================================
# 6. MAIN: N x 3 FIGURE
# ============================================================

def plot_comparison():

    ds_ref = load_dataset(REFERENCE, MONTH, DATA_TYPE)
    ds_exp = load_dataset(EXPERIMENT, MONTH, DATA_TYPE)

    n = len(SELECTED_POINTS)

    fig, axes = plt.subplots(
        nrows=n,
        ncols=3,
        figsize=(17, 3.7 * n + 1.0),
        sharex=True,
        squeeze=False,
        constrained_layout=True
    )

    # Extract and align data for all stations.
    all_data = []

    for station in SELECTED_POINTS:

        ref_da = extract_time_depth(
            ds_ref, VARIABLE, station
        )
        exp_da = extract_time_depth(
            ds_exp, VARIABLE, station
        )

        ref_da, exp_da, diff_da = align_experiments(
            ref_da, exp_da
        )

        all_data.append(
            (station, ref_da, exp_da, diff_da)
        )

    # Shared color limits for reference and experiment.
    combined = np.concatenate([
        da.values.ravel()
        for _, ref, exp, _ in all_data
        for da in (ref, exp)
    ])
    combined = combined[np.isfinite(combined)]

    if combined.size == 0:
        raise ValueError("No valid data to plot.")

    vmin = VMIN if VMIN is not None else np.min(combined)
    vmax = VMAX if VMAX is not None else np.max(combined)

    # Symmetric difference color limits.
    diffs = np.concatenate([
        diff.values.ravel()
        for _, _, _, diff in all_data
    ])
    diffs = diffs[np.isfinite(diffs)]

    if diffs.size == 0:
        raise ValueError("No valid difference data.")

    diff_max = (
        DIFF_MAX
        if DIFF_MAX is not None
        else np.max(np.abs(diffs))
    )

    # Avoid zero-width color range.
    if vmin == vmax:
        vmax = vmin + 1e-6

    if diff_max == 0:
        diff_max = 1e-6

    diff_norm = TwoSlopeNorm(
        vmin=-diff_max,
        vcenter=0,
        vmax=diff_max
    )

    # Plot each station.
    for i, (station, ref, exp, diff) in enumerate(all_data):

        # Reference: white contours
        m_ref = plot_time_depth(
            axes[i, 0],
            ref,
            CMAP,
            vmin,
            vmax,
            contour_levels=CONTOUR_LEVELS,
            contour_color=CONTOUR_COLOR
        )

        # Experiment: white contours
        m_exp = plot_time_depth(
            axes[i, 1],
            exp,
            CMAP,
            vmin,
            vmax,
            contour_levels=CONTOUR_LEVELS,
            contour_color=CONTOUR_COLOR
        )

        # Difference: black contours
        m_diff = plot_time_depth(
            axes[i, 2],
            diff,
            DIFF_CMAP,
            None,
            None,
            contour_levels=DIFF_CONTOUR_LEVELS,
            norm=diff_norm,
            contour_color=DIFF_CONTOUR_COLOR
        )

        # Same depth range across the three panels.
        max_depth = float(ref.deptht.max())

        for j in range(3):
            axes[i, j].set_ylim(max_depth, 0)

        axes[i, 0].set_ylabel(
            f"{station}\nDepth (m)",
            fontsize=10
        )

    # Column titles
    axes[0, 0].set_title(
        REFERENCE, fontsize=12, fontweight="bold"
    )
    axes[0, 1].set_title(
        EXPERIMENT, fontsize=12, fontweight="bold"
    )
    axes[0, 2].set_title(
        f"{EXPERIMENT} − {REFERENCE}",
        fontsize=12,
        fontweight="bold"
    )

    # Bottom x labels
    for j in range(3):
        axes[-1, j].set_xlabel("Time", fontsize=11)

    # Two colorbars:
    # 1. shared by reference and experiment
    # 2. difference
    # unit = UNITS.get(VARIABLE, "")

    cbar_main = fig.colorbar(
        m_ref,
        ax=axes[:, 0:2].ravel().tolist(),
        orientation="horizontal",
        fraction=0.035,
        pad=0.10,
        aspect=45
    )
    cbar_main.set_label(
        f"{VARIABLE} ", fontsize=10
    )

    cbar_diff = fig.colorbar(
        m_diff,
        ax=axes[:, 2].ravel().tolist(),
        orientation="horizontal",
        fraction=0.035,
        pad=0.10,
        aspect=22
    )
    cbar_diff.set_label(
        f"Difference ", fontsize=10
    )

    fig.suptitle(
        f"{VARIABLE}, "
        f"{EXPERIMENT} vs {REFERENCE}",
        fontsize=15,
        fontweight="bold"
    )



    plt.show()

    ds_ref.close()
    ds_exp.close()

    return fig, axes


# ============================================================
# 7. RUN
# ============================================================

fig, axes = plot_comparison()
```

```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.colors import TwoSlopeNorm
from pathlib import Path

# ============================================================
# 1. USER SETTINGS
# ============================================================

MONTH = "Mar"
EXPERIMENT = "CaSR+HRDPS2.5_wind"       # HRDPS1 / CaSR / HRDPS10 / "HRDPS2.5+CaSR_wind" / "CaSR+HRDPS2.5_wind"
REFERENCE = "HRDPS2.5"

VARIABLE = "diatoms"       # votemper / vosaline / nitrate 
DATA_TYPE = "Biological"      # Physical / Biological

# Start with one station; add more stations later.
SELECTED_POINTS = [
    # "Central SoG",
    # "Juan de Fuca",
    # "Haro Strait",
    # "Northern SoG",
    "Puget Sound",
    "Pam Rocks",
    "SandHeads",
]

# Color settings for the two experiments
CMAP = "viridis"
VMIN = None                 # e.g. 6
VMAX = None                 # e.g. 10

# Difference = experiment - reference
DIFF_CMAP = "RdBu_r"
DIFF_MAX = None             # e.g. 1.0, gives [-1, 1]

# ------------------------------------------------------------
# Contour levels: manually fill these in.
# None = no contour lines
# ------------------------------------------------------------

# CONTOUR_LEVELS = None
# Example:
CONTOUR_LEVELS = np.arange(0, 8, 2)
# CONTOUR_LEVELS = [20, 24, 28, 29, 30, 30.2, 30.4]

# DIFF_CONTOUR_LEVELS = None
# Example:
# DIFF_CONTOUR_LEVELS = np.arange(-1.0, 1.01, 0.25)
DIFF_CONTOUR_LEVELS = np.arange(-4, 4, 1)

SHOW_CONTOUR_LABELS = True
CONTOUR_COLOR = "white"
DIFF_CONTOUR_COLOR = "black"
CONTOUR_LINEWIDTH = 0.6

# Units (change as needed)
# UNITS = {
#     "votemper": "°C",
#     "vosaline": "g kg$^{-1}$",
#     "nitrate": "mmol m$^{-3}$",
#     "diatoms": "mmol N m$^{-3}$",
#     "flagellates": "mmol N m$^{-3}$",
# }


# ============================================================
# 2. LOAD DATA
# ============================================================

def get_file_path(experiment, month, data_type):
    prefix = EXPERIMENTS_head[experiment][month]
    suffix = EXPERIMENTS_tail[data_type]
    path = Path(prefix + suffix)

    if not path.is_file():
        raise FileNotFoundError(f"File not found: {path}")

    return path


def load_dataset(experiment, month, data_type):
    path = get_file_path(experiment, month, data_type)
    print(f"Opening: {path}")
    return xr.open_dataset(path)


# ============================================================
# 3. EXTRACT TIME-DEPTH DATA AT ONE GRID POINT
# ============================================================

def extract_time_depth(ds, variable, station):
    """
    Return a DataArray with dimensions:
        (time_counter, deptht)

    The original variable is expected to have dimensions:
        (time_counter, deptht, y, x)
    """

    y, x = POINTS[station]
    #n_layers = Layers[station]
    n_layers = min(20, Layers[station])

    da = ds[variable].isel(y=y, x=x)

    # Limit to valid depth levels.
    da = da.isel(deptht=slice(0, n_layers))

    # Apply ocean mask.
    point_mask = tmask.isel(y=y, x=x)

    # NEMO mesh masks often have a singleton time dimension.
    if "t" in point_mask.dims:
        point_mask = point_mask.isel(t=0)
    if "time_counter" in point_mask.dims:
        point_mask = point_mask.isel(time_counter=0)

    # Rename mask depth dimension if necessary.
    if "z" in point_mask.dims:
        point_mask = point_mask.rename({"z": "deptht"})

    point_mask = point_mask.isel(
        deptht=slice(0, n_layers)
    )

    # Use matching depth coordinates to avoid xarray misalignment.
    point_mask = point_mask.assign_coords(
        deptht=da.deptht
    )

    da = da.where(point_mask == 1)

    # Remove missing and invalid values.
    da = da.where(np.isfinite(da))

    # Ensure time-depth dimension order.
    da = da.transpose("time_counter", "deptht")

    return da.load()


# ============================================================
# 4. ALIGN TWO EXPERIMENTS
# ============================================================

def align_experiments(ref_da, exp_da):
    """
    Match common dates and depth levels.

    Useful because HRDPS1 March ends on March 28,
    while HRDPS2.5 March ends on March 31.
    """

    ref_da, exp_da = xr.align(
        ref_da,
        exp_da,
        join="inner"
    )

    if ref_da.sizes["time_counter"] == 0:
        raise ValueError("No overlapping time coordinates.")

    if ref_da.sizes["deptht"] == 0:
        raise ValueError("No overlapping depth coordinates.")

    diff_da = exp_da - ref_da

    return ref_da, exp_da, diff_da


# ============================================================
# 5. PLOT SINGLE TIME-DEPTH PANEL
# ============================================================

def plot_time_depth(
    ax,
    da,
    cmap,
    vmin,
    vmax,
    contour_levels=None,
    norm=None,
    contour_color="white"
):
    time = pd.to_datetime(da.time_counter.values)
    depth = da.deptht.values
    values = da.transpose("deptht", "time_counter").values

    # Pseudocolor (time-depth field)
    mesh = ax.pcolormesh(
        time,
        depth,
        values,
        shading="auto",
        cmap=cmap,
        vmin=vmin if norm is None else None,
        vmax=vmax if norm is None else None,
        norm=norm,
        rasterized=True
    )

    # Contour lines
    if contour_levels is not None and len(time) >= 2:
        cs = ax.contour(
            time,
            depth,
            values,
            levels=contour_levels,
            colors=contour_color,
            linewidths=CONTOUR_LINEWIDTH
        )

    if SHOW_CONTOUR_LABELS:
        ax.clabel(
            cs,
            inline=True,
            fontsize=7,
            fmt="%g",
            colors=contour_color
        )
    # Depth increases downward.
    ax.invert_yaxis()

    ax.xaxis.set_major_locator(
        mdates.AutoDateLocator(minticks=3, maxticks=7)
    )
    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%b %d")
    )

    ax.tick_params(axis="both", labelsize=9)
    ax.grid(False)

    return mesh


# ============================================================
# 6. MAIN: N x 3 FIGURE
# ============================================================

def plot_comparison():

    ds_ref = load_dataset(REFERENCE, MONTH, DATA_TYPE)
    ds_exp = load_dataset(EXPERIMENT, MONTH, DATA_TYPE)

    n = len(SELECTED_POINTS)

    fig, axes = plt.subplots(
        nrows=n,
        ncols=3,
        figsize=(17, 3.7 * n + 1.0),
        sharex=True,
        squeeze=False,
        constrained_layout=True
    )

    # Extract and align data for all stations.
    all_data = []

    for station in SELECTED_POINTS:

        ref_da = extract_time_depth(
            ds_ref, VARIABLE, station
        )
        exp_da = extract_time_depth(
            ds_exp, VARIABLE, station
        )

        ref_da, exp_da, diff_da = align_experiments(
            ref_da, exp_da
        )

        all_data.append(
            (station, ref_da, exp_da, diff_da)
        )

    # Shared color limits for reference and experiment.
    combined = np.concatenate([
        da.values.ravel()
        for _, ref, exp, _ in all_data
        for da in (ref, exp)
    ])
    combined = combined[np.isfinite(combined)]

    if combined.size == 0:
        raise ValueError("No valid data to plot.")

    vmin = VMIN if VMIN is not None else np.min(combined)
    vmax = VMAX if VMAX is not None else np.max(combined)

    # Symmetric difference color limits.
    diffs = np.concatenate([
        diff.values.ravel()
        for _, _, _, diff in all_data
    ])
    diffs = diffs[np.isfinite(diffs)]

    if diffs.size == 0:
        raise ValueError("No valid difference data.")

    diff_max = (
        DIFF_MAX
        if DIFF_MAX is not None
        else np.max(np.abs(diffs))
    )

    # Avoid zero-width color range.
    if vmin == vmax:
        vmax = vmin + 1e-6

    if diff_max == 0:
        diff_max = 1e-6

    diff_norm = TwoSlopeNorm(
        vmin=-diff_max,
        vcenter=0,
        vmax=diff_max
    )

    # Plot each station.
    for i, (station, ref, exp, diff) in enumerate(all_data):

        # Reference: white contours
        m_ref = plot_time_depth(
            axes[i, 0],
            ref,
            CMAP,
            vmin,
            vmax,
            contour_levels=CONTOUR_LEVELS,
            contour_color=CONTOUR_COLOR
        )

        # Experiment: white contours
        m_exp = plot_time_depth(
            axes[i, 1],
            exp,
            CMAP,
            vmin,
            vmax,
            contour_levels=CONTOUR_LEVELS,
            contour_color=CONTOUR_COLOR
        )

        # Difference: black contours
        m_diff = plot_time_depth(
            axes[i, 2],
            diff,
            DIFF_CMAP,
            None,
            None,
            contour_levels=DIFF_CONTOUR_LEVELS,
            norm=diff_norm,
            contour_color=DIFF_CONTOUR_COLOR
        )

        # Same depth range across the three panels.
        max_depth = float(ref.deptht.max())

        for j in range(3):
            axes[i, j].set_ylim(max_depth, 0)

        axes[i, 0].set_ylabel(
            f"{station}\nDepth (m)",
            fontsize=10
        )

    # Column titles
    axes[0, 0].set_title(
        REFERENCE, fontsize=12, fontweight="bold"
    )
    axes[0, 1].set_title(
        EXPERIMENT, fontsize=12, fontweight="bold"
    )
    axes[0, 2].set_title(
        f"{EXPERIMENT} − {REFERENCE}",
        fontsize=12,
        fontweight="bold"
    )

    # Bottom x labels
    for j in range(3):
        axes[-1, j].set_xlabel("Time", fontsize=11)

    # Two colorbars:
    # 1. shared by reference and experiment
    # 2. difference
    # unit = UNITS.get(VARIABLE, "")

    cbar_main = fig.colorbar(
        m_ref,
        ax=axes[:, 0:2].ravel().tolist(),
        orientation="horizontal",
        fraction=0.035,
        pad=0.10,
        aspect=45
    )
    cbar_main.set_label(
        f"{VARIABLE} ", fontsize=10
    )

    cbar_diff = fig.colorbar(
        m_diff,
        ax=axes[:, 2].ravel().tolist(),
        orientation="horizontal",
        fraction=0.035,
        pad=0.10,
        aspect=22
    )
    cbar_diff.set_label(
        f"Difference ", fontsize=10
    )

    fig.suptitle(
        f"{VARIABLE}, "
        f"{EXPERIMENT} vs {REFERENCE}",
        fontsize=15,
        fontweight="bold"
    )



    plt.show()

    ds_ref.close()
    ds_exp.close()

    return fig, axes


# ============================================================
# 7. RUN
# ============================================================

fig, axes = plot_comparison()
```

```python

import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pathlib import Path


# ============================================================
# 1. USER SETTINGS
# ============================================================

MONTH = "Mar"
N_LAYERS = 20

SELECTED_POINTS = [
    "Puget Sound",
    "Pam Rocks",
    "SandHeads",
]

MODELS = [
    "HRDPS2.5",
    "CaSR",
    "HRDPS2.5+CaSR_wind",
    "CaSR+HRDPS2.5_wind",
]

PHYSICAL_VARS = [
    ("votemper", "Temperature"),
    ("vosaline", "Salinity"),
]

BIOLOGICAL_VARS = [
    ("nitrate", "Nitrate"),
    ("diatoms", "Diatoms"),
    ("flagellates", "Flagellates"),
]



COLORS = {
    "HRDPS2.5": "black",
    "CaSR": "tab:blue",
    "HRDPS2.5+CaSR_wind": "tab:red",
    "CaSR+HRDPS2.5_wind": "tab:green",
}

LINESTYLES = {
    "HRDPS2.5": "-",
    "CaSR": "-",
    "HRDPS2.5+CaSR_wind": "--",
    "CaSR+HRDPS2.5_wind": "--",
}

SAVE_DIR = Path("./top20_timeseries")
SAVE_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. LOAD DATA
# ============================================================

def get_file_path(experiment, month, data_type):

    prefix = EXPERIMENTS_head[experiment][month]
    suffix = EXPERIMENTS_tail[data_type]

    path = Path(prefix + suffix)

    if not path.is_file():
        raise FileNotFoundError(
            f"File not found: {path}"
        )

    return path


def load_datasets(data_type):

    datasets = {}

    for model in MODELS:

        path = get_file_path(
            model, MONTH, data_type
        )

        print(f"Opening: {path}")

        datasets[model] = xr.open_dataset(path)

    return datasets


# ============================================================
# 3. EXTRACT FIRST 20 LAYERS AND CALCULATE SIMPLE MEAN
# ============================================================

def extract_top20_mean(ds, variable, station):

    """
    Arithmetic mean over the first 20 vertical layers.

    Input:
        (time_counter, deptht, y, x)

    Output:
        (time_counter,)

    No thickness weighting.
    """

    y, x = POINTS[station]

    n_layers = min(
        N_LAYERS,
        Layers[station]
    )

    # Extract selected point
    da = ds[variable].isel(
        y=y,
        x=x,
        deptht=slice(0, n_layers)
    )

    # Extract mask
    point_mask = tmask.isel(
        y=y,
        x=x
    )

    # Remove time dimension from mask
    if "t" in point_mask.dims:
        point_mask = point_mask.isel(t=0)

    if "time_counter" in point_mask.dims:
        point_mask = point_mask.isel(time_counter=0)

    # Rename vertical dimension
    if "z" in point_mask.dims:
        point_mask = point_mask.rename(
            {"z": "deptht"}
        )

    point_mask = point_mask.isel(
        deptht=slice(0, n_layers)
    )

    # Match depth coordinates
    point_mask = point_mask.assign_coords(
        deptht=da.deptht
    )

    # Apply ocean mask
    da = da.where(point_mask == 1)

    # Remove invalid values
    da = da.where(np.isfinite(da))

    # Simple arithmetic mean
    da_mean = da.mean(
        dim="deptht",
        skipna=True
    )

    return da_mean.load()


# ============================================================
# 4. GENERAL PLOTTING FUNCTION
# ============================================================

def plot_top20_timeseries(data_type, variables):

    datasets = load_datasets(data_type)

    nrows = len(SELECTED_POINTS)
    ncols = len(variables)

    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=(5.5 * ncols, 3.2 * nrows),
        sharex=True,
        squeeze=False
    )

    try:

        # ----------------------------------------------------
        # Loop over stations
        # ----------------------------------------------------

        for row, station in enumerate(SELECTED_POINTS):

            # ------------------------------------------------
            # Loop over variables
            # ------------------------------------------------

            for col, (variable, var_title) in enumerate(variables):

                ax = axes[row, col]

                # --------------------------------------------
                # Loop over experiments
                # --------------------------------------------

                for model in MODELS:

                    da_mean = extract_top20_mean(
                        datasets[model],
                        variable,
                        station
                    )

                    time = pd.to_datetime(
                        da_mean.time_counter.values
                    )

                    ax.plot(
                        time,
                        da_mean.values,
                        color=COLORS[model],
                        linestyle=LINESTYLES[model],
                        linewidth=1.6,
                        label=model
                    )

                # --------------------------------------------
                # Column titles
                # --------------------------------------------

                if row == 0:
                    ax.set_title(
                        var_title,
                        fontsize=13,
                        fontweight="bold"
                    )

                # --------------------------------------------
                # Station name on left
                # --------------------------------------------

                if col == 0:
                    ax.set_ylabel(
                        station,
                        fontsize=12,
                        fontweight="bold"
                    )

                # --------------------------------------------
                # Grid and ticks
                # --------------------------------------------

                ax.grid(
                    True,
                    alpha=0.25
                )

                ax.tick_params(
                    axis="both",
                    labelsize=9
                )


                # Date formatting
                ax.xaxis.set_major_locator(
                    mdates.DayLocator(interval=5)
                )

                ax.xaxis.set_major_formatter(
                    mdates.DateFormatter("%b %d")
                )

        # ----------------------------------------------------
        # Bottom x-axis labels
        # ----------------------------------------------------

        for col in range(ncols):
            axes[-1, col].set_xlabel(
                "Time",
                fontsize=11
            )

        # ----------------------------------------------------
        # One shared legend
        # ----------------------------------------------------

        handles, labels = axes[0, 0].get_legend_handles_labels()

        fig.legend(
            handles,
            labels,
            loc="upper center",
            ncol=4,
            bbox_to_anchor=(0.5, 0.94),
            frameon=False,
            fontsize=10
        )

        # ----------------------------------------------------
        # Figure title
        # ----------------------------------------------------

        fig.suptitle(
            f"{data_type} Variables - "
            f"Top {N_LAYERS} Layers in ({MONTH})",
            fontsize=15,
            fontweight="bold",
            y=0.995
        )

        plt.tight_layout(
            rect=[0, 0, 1, 0.90]
        )

        plt.show()

        return fig, axes

    finally:
        for ds in datasets.values():
            ds.close()


# ============================================================
# 5. PHYSICAL: 3 ROWS x 2 COLUMNS
# ============================================================

fig_phy, axes_phy = plot_top20_timeseries(
    data_type="Physical",
    variables=PHYSICAL_VARS
)


# ============================================================
# 6. BIOLOGICAL: 3 ROWS x 3 COLUMNS
# ============================================================

fig_bio, axes_bio = plot_top20_timeseries(
    data_type="Biological",
    variables=BIOLOGICAL_VARS
)
```
