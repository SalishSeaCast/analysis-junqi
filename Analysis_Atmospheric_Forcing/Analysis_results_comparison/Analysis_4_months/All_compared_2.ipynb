```python
# Skip HRDPS2.5 for now

List_exp = ['HRDPS1', 'CaSR', 'HRDPS10']

file_tail = EXPERIMENTS_tail['Physical']
```

```python
import numpy as np

for month in List_month:

    # Store results for each region
    results = {region: [] for region in REGIONS}

    for exp in List_exp:

        print(f'Processing {exp} - {month}')

        # Open T-grid file
        path_file = EXPERIMENTS_head[exp][month] + file_tail

        ds = xr.open_dataset(
            path_file,
            chunks={'time_counter': 7}
        )

        for region_name, (j0, j1, i0, i1) in REGIONS.items():

            print(f'  {region_name}')

            # Select region

            mask_region = tmask.isel(
                y=slice(j0, j1),
                x=slice(i0, i1)
            )

            area_region = area.isel(
                y=slice(j0, j1),
                x=slice(i0, i1)
            )

            # Salinity: mask land points first
            sal = ds['vosaline'].isel(
                y=slice(j0, j1),
                x=slice(i0, i1)
            ).where(mask_region == 1)

            # Layer thickness: mask land points first
            thickness = ds['e3t'].isel(
                y=slice(j0, j1),
                x=slice(i0, i1)
            ).where(mask_region == 1)

            # Area: use surface mask
            area_region = area_region.where(
                mask_region.isel(deptht=0) == 1
            )

            # --------------------------------
            # 1. Sea Surface Salinity (SSS)
            # --------------------------------

            sss = sal.isel(deptht=0)

            weight = area_region.where(sss.notnull())

            sss_mean = (
                (sss * weight).sum(dim=['y', 'x'])
                / weight.sum(dim=['y', 'x'])
            )

            # --------------------------------
            # 2. Regional salinity profile
            # --------------------------------

            weight = area_region.where(sal.notnull())

            sal_mean = (
                (sal * weight).sum(dim=['y', 'x'])
                / weight.sum(dim=['y', 'x'])
            )

            weight_e3t = area_region.where(thickness.notnull())

            e3t_mean = (
                (thickness * weight_e3t).sum(dim=['y', 'x'])
                / weight_e3t.sum(dim=['y', 'x'])
            )

            # --------------------------------
            # 3. Halocline
            # --------------------------------

            depth_values = sal_mean.deptht.values

            depth_mid = (
                depth_values[:-1] + depth_values[1:]
            ) / 2

            dsal = sal_mean.diff('deptht')
            dsal = dsal.assign_coords(deptht=depth_mid)

            dz = e3t_mean.isel(deptht=slice(0, -1))
            dz = dz.assign_coords(deptht=depth_mid)

            gradient = abs(dsal / dz)
            gradient = gradient.where(np.isfinite(gradient))

            halocline_strength = gradient.max(dim='deptht')

            halocline_depth = gradient.deptht.isel(
                deptht=gradient.fillna(-np.inf).argmax(dim='deptht').compute()
            )

            halocline_depth = halocline_depth.where(
                gradient.notnull().any(dim='deptht')
            )

            # --------------------------------
            # 4. Save to dataframe
            # --------------------------------

            df = pd.DataFrame({
                'date': pd.to_datetime(ds.time_counter.values).normalize(),
                f'{exp}__SSS': sss_mean.compute().values,
                f'{exp}__halocline_depth_m': halocline_depth.compute().values,
                f'{exp}__halocline_strength_psu_per_m': halocline_strength.compute().values
            })

            results[region_name].append(df)

        ds.close()

    # --------------------------------
    # 5. Save CSV for each region
    # --------------------------------

    for region_name, frames in results.items():

        df = frames[0]

        for frame in frames[1:]:
            df = df.merge(frame, on='date', how='outer')

        j0, j1, i0, i1 = REGIONS[region_name]

        df['region'] = region_name
        df['j0'] = j0
        df['j1'] = j1
        df['i0'] = i0
        df['i1'] = i1

        df = df.sort_values('date')

        filename = region_name.replace(' ', '_')

        df.to_csv(
            path_output + f'{filename}_{month}_physical.csv',
            index=False
        )

        print(f'Saved: {filename}_{month}_physical.csv')

print('Physical extraction finished!')
```

```python
# Rework Parallel Computation


import numpy as np
import pandas as pd
import xarray as xr
import dask

# --------------------------------
# 1. Dask 并行设置
# --------------------------------

N_WORKERS = 8  # 根据 CPU 核心数和内存调整
TIME_CHUNK = 14

# 预先准备各个区域的 mask 和 area
region_info = {}

for region_name, (j0, j1, i0, i1) in REGIONS.items():

    mask_r = tmask.isel(
        y=slice(j0, j1),
        x=slice(i0, i1)
    )

    area_r = area.isel(
        y=slice(j0, j1),
        x=slice(i0, i1)
    ).where(mask_r.isel(deptht=0) == 1)

    region_info[region_name] = {
        "bounds": (j0, j1, i0, i1),
        "mask": mask_r,
        "area": area_r,
    }


# --------------------------------
# 2. 定义单个 region 的计算任务
# 不在函数内部调用 compute()
# --------------------------------

def calculate_region(ds, region_name):

    info = region_info[region_name]

    j0, j1, i0, i1 = info["bounds"]
    mask_r = info["mask"]
    area_r = info["area"]

    ds_r = ds.isel(
        y=slice(j0, j1),
        x=slice(i0, i1)
    )

    sal = ds_r["vosaline"].where(mask_r == 1)
    thickness = ds_r["e3t"].where(mask_r == 1)

    # 1. Sea Surface Salinity
    sss = sal.isel(deptht=0)

    weight_sss = area_r.where(sss.notnull())

    sss_mean = (
        (sss * weight_sss).sum(dim=["y", "x"])
        / weight_sss.sum(dim=["y", "x"])
    )

    # 2. Regional salinity profile
    weight_sal = area_r.where(sal.notnull())

    sal_mean = (
        (sal * weight_sal).sum(dim=["y", "x"])
        / weight_sal.sum(dim=["y", "x"])
    )

    weight_e3t = area_r.where(thickness.notnull())

    e3t_mean = (
        (thickness * weight_e3t).sum(dim=["y", "x"])
        / weight_e3t.sum(dim=["y", "x"])
    )

    # 3. Halocline
    depth_values = sal_mean.deptht.values

    depth_mid = (
        depth_values[:-1] + depth_values[1:]
    ) / 2

    dsal = sal_mean.diff("deptht")
    dsal = dsal.assign_coords(deptht=depth_mid)

    dz = e3t_mean.isel(deptht=slice(0, -1))
    dz = dz.assign_coords(deptht=depth_mid)

    gradient = abs(dsal / dz)
    gradient = gradient.where(np.isfinite(gradient))

    halocline_strength = gradient.max(dim="deptht")

    # 保持延迟计算，不提前 compute()
    max_idx = gradient.fillna(-np.inf).argmax(dim="deptht")

    halocline_depth = gradient.deptht.isel(
        deptht=max_idx
    ).where(
        gradient.notnull().any(dim="deptht")
    )

    # 4. 返回延迟计算的 Dataset
    return xr.Dataset({
        "SSS": sss_mean,
        "halocline_depth_m": halocline_depth,
        "halocline_strength_psu_per_m": halocline_strength
    })


# --------------------------------
# 3. Main loop
# --------------------------------

for month in List_month:

    results = {region: [] for region in REGIONS}

    for exp in List_exp:

        print(f"Processing {exp} - {month}")

        path_file = EXPERIMENTS_head[exp][month] + file_tail

        with xr.open_dataset(
            path_file,
            chunks={"time_counter": TIME_CHUNK}
        ) as ds:

            # 创建所有区域的延迟计算任务
            region_names = list(REGIONS)

            tasks = [
                calculate_region(ds, region_name)
                for region_name in region_names
            ]

            # 一次提交所有区域的计算
            computed_results = dask.compute(
                *tasks,
                scheduler="threads",
                num_workers=N_WORKERS
            )

            # 所有计算结束后，再生成 DataFrame
            for region_name, result in zip(
                region_names, computed_results
            ):

                df = pd.DataFrame({
                    "date": pd.to_datetime(
                        result.time_counter.values
                    ).normalize(),

                    f"{exp}__SSS":
                        result["SSS"].values,

                    f"{exp}__halocline_depth_m":
                        result["halocline_depth_m"].values,

                    f"{exp}__halocline_strength_psu_per_m":
                        result["halocline_strength_psu_per_m"].values,
                })

                results[region_name].append(df)

    # --------------------------------
    # 4. Save CSV for each region
    # --------------------------------

    for region_name, frames in results.items():

        df = frames[0]

        for frame in frames[1:]:
            df = df.merge(
                frame, on="date", how="outer"
            )

        j0, j1, i0, i1 = REGIONS[region_name]

        df["region"] = region_name
        df["j0"] = j0
        df["j1"] = j1
        df["i0"] = i0
        df["i1"] = i1

        df = df.sort_values("date")

        filename = region_name.replace(" ", "_")

        df.to_csv(
            path_output + f"{filename}_{month}_physical.csv",
            index=False
        )

        print(f"Saved: {filename}_{month}_physical.csv")

print("Physical extraction finished!")
```

```python

from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd
import xarray as xr

from dask import compute
from dask.distributed import Client, LocalCluster

warnings.filterwarnings("ignore", category=UserWarning)


# ============================================================
# 1. SETTINGS
# ============================================================

OUTPUT_DIR = Path(
    "/home/jqiu/analysis-junqi/"
    "Analysis_Atmospheric_Forcing/"
    "Analysis_results_comparison/"
    "Analysis_4_months/Data_4_months/physical"
)

# Shared grid files
MASK_FILE = Path(
    "/ocean/jqiu/Grid/mesh_mask202108.nc"
)

AREA_FILE = Path(
    "/ocean/jqiu/results/Flux/flux_00mar23_CaSR/"
    "SalishSea_1d_20230301_20230331_biol_T.nc"
)

REGIONS = {
    "Central SoG":  (450, 500, 250, 300),
    "Juan de Fuca": (300, 365,  50, 100),
    "Haro Strait":  (280, 320, 210, 250),
    "Northern SoG": (650, 700, 140, 180),
    "Puget Sound":  (75,  125, 225, 280),
    "Pam Rocks":    (490, 510, 330, 350),
    "SandHeads":    (420, 440, 280, 300),
}

MONTHS = ["Mar", "Jun", "Sep", "Dec"]

# Dask
N_WORKERS = 1
THREADS_PER_WORKER = 4
USE_PROCESSES = False

SPATIAL_CHUNK_Y = 128
SPATIAL_CHUNK_X = 128
T_TIME_CHUNK = 7

# Three available experiments.
# HRDPS2.5 excluded until results are ready.
EXPERIMENTS_HEAD = {
    "HRDPS1": {
        "Mar": (
            "/ocean/jqiu/results/Flux/flux_00mar23_hrdps1/"
            "SalishSea_1d_20230301_20230328"
        ),
        "Jun": (
            "/ocean/jqiu/results/Flux/flux_00jun23_hrdps1/"
            "SalishSea_1d_20230601_20230630"
        ),
        "Sep": (
            "/ocean/jqiu/results/Flux/flux_00sep23_hrdps1/"
            "SalishSea_1d_20230901_20230930"
        ),
        "Dec": (
            "/ocean/jqiu/results/Flux/flux_00dec23_hrdps1/"
            "SalishSea_1d_20231201_20231231"
        ),
    },

    "CaSR": {
        "Mar": (
            "/ocean/jqiu/results/Flux/flux_00mar23_CaSR/"
            "SalishSea_1d_20230301_20230331"
        ),
        "Jun": (
            "/ocean/jqiu/results/Flux/flux_00jun23_CaSR/"
            "SalishSea_1d_20230601_20230630"
        ),
        "Sep": (
            "/ocean/jqiu/results/Flux/flux_00sep23_CaSR/"
            "SalishSea_1d_20230901_20230930"
        ),
        "Dec": (
            "/ocean/jqiu/results/Flux/flux_00dec23_CaSR/"
            "SalishSea_1d_20231201_20231231"
        ),
    },

    "HRDPS10": {
        "Mar": (
            "/ocean/jqiu/results/Flux/flux_00mar23_hrdps10/"
            "SalishSea_1d_20230301_20230331"
        ),
        "Jun": (
            "/ocean/jqiu/results/Flux/flux_00jun23_hrdps10/"
            "SalishSea_1d_20230601_20230630"
        ),
        "Sep": (
            "/ocean/jqiu/results/Flux/flux_00sep23_hrdps10/"
            "SalishSea_1d_20230901_20230930"
        ),
        "Dec": (
            "/ocean/jqiu/results/Flux/flux_00dec23_hrdps10/"
            "SalishSea_1d_20231201_20231231"
        ),
    },
}

PHYSICAL_TAIL = "_grid_T.nc"


# ============================================================
# 2. FILE AND GRID HELPERS
# ============================================================

def safe_filename(name):
    name = re.sub(r"\s+", "_", name.strip())
    name = re.sub(r"[^A-Za-z0-9_.-]+", "_", name)
    return name.strip("_")


def require_vars(ds, variables, label):
    missing = [
        v for v in variables
        if v not in ds.variables
    ]

    if missing:
        raise KeyError(
            f"{label} missing variables: {missing}\n"
            f"Available: {list(ds.variables)}"
        )


def get_physical_file(exp_name, month):
    return Path(
        EXPERIMENTS_HEAD[exp_name][month]
        + PHYSICAL_TAIL
    )


def validate_input_files():
    """Check input files before expensive processing."""
    files = [MASK_FILE, AREA_FILE]

    for exp_name in EXPERIMENTS_HEAD:
        for month in MONTHS:
            files.append(
                get_physical_file(exp_name, month)
            )

    missing = [
        str(p) for p in files
        if not p.is_file()
    ]

    if missing:
        raise FileNotFoundError(
            "Missing input files:\n"
            + "\n".join(missing)
        )

    print("All required input files found.")


def load_shared_grids():
    """
    Load area and 3-D tmask once into memory.

    area:  (y, x)
    tmask: (deptht, y, x)
    """

    print("\nLoading shared area...")

    with xr.open_dataset(AREA_FILE) as ds:
        require_vars(ds, ["area"], "Area source")

        area = (
            ds["area"]
            .squeeze(drop=True)
            .transpose("y", "x")
            .load()
        )

    print("Loading shared tmask...")

    with xr.open_dataset(MASK_FILE) as ds:
        require_vars(ds, ["tmask"], "Mesh mask")

        tmask = (
            ds["tmask"]
            .isel(t=0, drop=True)
            .rename({"z": "deptht"})
            .transpose("deptht", "y", "x")
            .load()
        )

    # Check common horizontal grid shape.
    for dim in ("y", "x"):
        if area.sizes[dim] != tmask.sizes[dim]:
            raise ValueError(
                f"Grid mismatch on {dim}: "
                f"area={area.sizes[dim]}, "
                f"tmask={tmask.sizes[dim]}"
            )

    # Keep numeric arrays with explicit dimension names,
    # but without inherited coordinate labels.
    # This prevents accidental coordinate alignment.
    area = xr.DataArray(
        area.data,
        dims=("y", "x"),
    )

    tmask = xr.DataArray(
        tmask.data,
        dims=("deptht", "y", "x"),
    )

    # Invalid or non-positive area -> NaN.
    area = area.where(
        np.isfinite(area) & (area > 0)
    )

    # Boolean mask: True = valid water cell.
    tmask = tmask == 1

    print(
        f"Area shape: {area.shape}, "
        f"tmask shape: {tmask.shape}"
    )
    print("Shared grids loaded.")

    return area, tmask


def build_region_grids(area, tmask):
    """
    Precompute regional views of shared grids.
    No additional file reading.
    """

    area_by_region = {}
    mask_by_region = {}

    for region_name, region in REGIONS.items():

        j0, j1, i0, i1 = region

        if not (
            0 <= j0 < j1 <= area.sizes["y"]
            and 0 <= i0 < i1 <= area.sizes["x"]
        ):
            raise ValueError(
                f"Invalid region bounds: {region_name}"
            )

        area_by_region[region_name] = area.isel(
            y=slice(j0, j1),
            x=slice(i0, i1),
        )

        mask_by_region[region_name] = tmask.isel(
            y=slice(j0, j1),
            x=slice(i0, i1),
        )

    return area_by_region, mask_by_region


def open_physical_t(path, label):
    """
    Open one experiment-month dataset lazily.
    """

    if not path.is_file():
        raise FileNotFoundError(
            f"{label}: {path}"
        )

    chunks = {
        "time_counter": T_TIME_CHUNK,
        "y": SPATIAL_CHUNK_Y,
        "x": SPATIAL_CHUNK_X,
    }

    return xr.open_dataset(
        path,
        chunks=chunks,
        drop_variables=["area", "e3t"],
    )


# ============================================================
# 3. PHYSICAL CALCULATIONS
# ============================================================

def area_weighted_mean(da, area):
    """
    Area-weighted mean over y, x.

    Only finite data with valid positive area
    contribute to the mean.
    """

    valid = np.isfinite(da) & np.isfinite(area)

    weights = area.where(valid)

    numerator = (
        da.where(valid) * weights
    ).sum(
        dim=("y", "x"),
        skipna=True,
    )

    denominator = weights.sum(
        dim=("y", "x"),
        skipna=True,
    )

    return (numerator / denominator).where(
        denominator > 0
    )


def halocline_from_region(salinity, area):
    """
    Calculate halocline from regional mean salinity.

    1. Area-average salinity at each depth.
    2. Calculate adjacent-level salinity differences.
    3. Divide by adjacent deptht differences.
    4. Find maximum absolute vertical gradient.

    Returns:
        halocline_depth (m)
        halocline_strength (psu/m)
    """

    sal_profile = area_weighted_mean(
        salinity,
        area,
    )

    # Small fixed depth coordinate -> NumPy.
    depth = np.asarray(
        sal_profile["deptht"].values,
        dtype=np.float64,
    )

    if depth.ndim != 1 or depth.size < 2:
        raise ValueError(
            "Need at least two depth levels."
        )

    dz = np.diff(depth)

    if not np.all(
        np.isfinite(dz) & (dz > 0)
    ):
        raise ValueError(
            "deptht must be finite and "
            "strictly increasing."
        )

    # Midpoint between adjacent T-grid centers.
    interface_depth = 0.5 * (
        depth[:-1] + depth[1:]
    )

    # Delta S along deptht.
    d_sal = sal_profile.diff("deptht")

    d_sal = d_sal.assign_coords(
        deptht=interface_depth
    )

    # Numeric depth spacing, with explicit depth labels.
    d_depth = xr.DataArray(
        dz,
        dims=("deptht",),
        coords={"deptht": interface_depth},
    )

    # |dS/dz| [psu/m].
    gradient = np.abs(d_sal / d_depth)

    gradient = gradient.where(
        np.isfinite(gradient)
    )

    has_valid = gradient.notnull().any(
        dim="deptht"
    )

    safe_gradient = gradient.fillna(-np.inf)

    max_index = safe_gradient.argmax(
        dim="deptht"
    )

    strength = gradient.max(
        dim="deptht",
        skipna=True,
    )

    halo_depth = gradient["deptht"].isel(
        deptht=max_index
    )

    return (
        halo_depth.where(has_valid),
        strength.where(has_valid),
    )


def build_physical_arrays(
    ds_t,
    area_t,
    tmask_t,
    region,
    exp_name,
):
    """
    Calculate masked:
        SSS
        Halocline depth
        Halocline strength
    """

    j0, j1, i0, i1 = region

    require_vars(
        ds_t,
        ["vosaline", "deptht", "time_counter"],
        f"{exp_name} T",
    )

    # Regional subset.
    t = ds_t.isel(
        y=slice(j0, j1),
        x=slice(i0, i1),
    )

    salinity = t["vosaline"]

    # Ensure expected dimensions.
    required_dims = {
        "time_counter", "deptht", "y", "x"
    }

    if set(salinity.dims) != required_dims:
        raise ValueError(
            f"Unexpected salinity dims: {salinity.dims}"
        )

    # Check mask / area shape before broadcasting.
    for dim in ("deptht", "y", "x"):
        if (
            tmask_t.sizes[dim]
            != salinity.sizes[dim]
        ):
            raise ValueError(
                f"{exp_name}: tmask mismatch "
                f"on dimension {dim}"
            )

    for dim in ("y", "x"):
        if (
            area_t.sizes[dim]
            != salinity.sizes[dim]
        ):
            raise ValueError(
                f"{exp_name}: area mismatch "
                f"on dimension {dim}"
            )

    # --------------------------------------------------
    # Apply the FULL 3-D mask before spatial averaging.
    #
    # tmask_t has dims (deptht, y, x), no coordinate
    # labels; broadcasting is by named dimension and
    # corresponding positional index.
    # --------------------------------------------------

    salinity = salinity.where(
        tmask_t & np.isfinite(salinity)
        & (salinity > 0)
    )

    # --------------------------------------------------
    # Surface salinity.
    # --------------------------------------------------

    sss = area_weighted_mean(
        salinity.isel(deptht=0),
        area_t,
    )

    # --------------------------------------------------
    # Halocline using deptht differences.
    # --------------------------------------------------

    halo_depth, halo_strength = (
        halocline_from_region(
            salinity,
            area_t,
        )
    )

    return [
        (f"{exp_name}__SSS", sss),
        (
            f"{exp_name}__halocline_depth_m",
            halo_depth,
        ),
        (
            f"{exp_name}__halocline_strength_psu_per_m",
            halo_strength,
        ),
    ]


# ============================================================
# 4. DATAFRAME HELPERS
# ============================================================

def da_to_daily_dataframe(da, column_name):
    """
    Convert computed 1-D time series to daily data.

    Multiple time records on the same date
    are averaged.
    """

    da = da.rename(column_name)

    df = da.to_dataframe().reset_index()

    df["date"] = pd.to_datetime(
        df["time_counter"]
    ).dt.normalize()

    return (
        df.groupby("date", as_index=False)[column_name]
        .mean()
    )


def merge_frames(frames):
    """
    Merge experiment/month series by date.

    Same variable from different months:
        concatenate vertically.

    Different variables/experiments:
        merge horizontally by date.
    """

    if not frames:
        return pd.DataFrame(columns=["date"])

    # Concatenate all series vertically.
    combined = pd.concat(
        frames,
        ignore_index=True,
    )

    # Combine rows with the same date.
    # Each column contains data only for its own
    # experiment/month, so first() selects the
    # first non-NaN value.
    out = (
        combined
        .groupby("date", as_index=False)
        .first()
    )

    return (
        out.sort_values("date")
        .reset_index(drop=True)
    )


def add_region_metadata(out, region_name, region):

    j0, j1, i0, i1 = region

    out.insert(
        1,
        "month",
        out["date"].dt.strftime("%b"),
    )

    out.insert(
        2,
        "region",
        region_name,
    )

    out["j0"] = j0
    out["j1"] = j1
    out["i0"] = i0
    out["i1"] = i1

    return out


# ============================================================
# 5. MAIN
# ============================================================

def main():

    OUTPUT_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )

    # Check all input files before starting Dask.
    validate_input_files()

    cluster = LocalCluster(
        n_workers=N_WORKERS,
        threads_per_worker=THREADS_PER_WORKER,
        processes=USE_PROCESSES,
        dashboard_address=":0",
    )

    client = Client(cluster)

    print("\nDask client started")
    print(f"Dashboard: {client.dashboard_link}")
    print(
        f"Workers={N_WORKERS}, "
        f"threads/worker={THREADS_PER_WORKER}, "
        f"processes={USE_PROCESSES}"
    )

    frames_by_region = {
        region_name: []
        for region_name in REGIONS
    }

    try:

        # ------------------------------------------------
        # Step 1: Load area and tmask ONLY ONCE.
        # ------------------------------------------------

        shared_area, shared_tmask = load_shared_grids()

        area_by_region, mask_by_region = (
            build_region_grids(
                shared_area,
                shared_tmask,
            )
        )

        # ------------------------------------------------
        # Step 2: Loop through experiments and months.
        # ------------------------------------------------

        for exp_name in EXPERIMENTS_HEAD:

            print(f"\n{'=' * 60}")
            print(f"Experiment: {exp_name}")
            print(f"{'=' * 60}")

            for month in MONTHS:

                file_path = get_physical_file(
                    exp_name,
                    month,
                )

                print(
                    f"\nOpening {exp_name} {month}: "
                    f"{file_path}"
                )

                # Open only ONE experiment-month.
                ds_t = open_physical_t(
                    file_path,
                    f"{exp_name} {month}",
                )

                try:

                    for region_name, region in REGIONS.items():

                        print(
                            f"  Processing {exp_name} "
                            f"{month}: {region_name}"
                        )

                        area_t = area_by_region[region_name]
                        tmask_t = mask_by_region[region_name]

                        lazy_outputs = build_physical_arrays(
                            ds_t,
                            area_t,
                            tmask_t,
                            region,
                            exp_name,
                        )

                        names = [
                            name for name, _ in lazy_outputs
                        ]

                        arrays = [
                            arr for _, arr in lazy_outputs
                        ]

                        # Share computation between the
                        # three outputs where possible.
                        computed = compute(*arrays)

                        frames_by_region[region_name].extend(
                            da_to_daily_dataframe(arr, name)
                            for name, arr in zip(names, computed)
                        )

                        del lazy_outputs, arrays, computed

                finally:
                    ds_t.close()

                    print(
                        f"Closed {exp_name} {month}."
                    )

        # ------------------------------------------------
        # Step 3: Save ONE CSV per region.
        # ------------------------------------------------

        print("\nWriting physical CSV files...")

        for region_name, region in REGIONS.items():

            out = merge_frames(
                frames_by_region[region_name]
            )

            out = add_region_metadata(
                out,
                region_name,
                region,
            )

            filename = (
                OUTPUT_DIR
                / f"{safe_filename(region_name)}_physical.csv"
            )

            out.to_csv(
                filename,
                index=False,
            )

            print(
                f"Saved -> {filename} "
                f"({len(out)} dates)"
            )

    finally:
        client.close()
        cluster.close()

    print("\nFour-month physical processing done.")


if __name__ == "__main__":
    main()
```
